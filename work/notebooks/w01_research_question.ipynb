## 3. Quick look at the data (2-3 real numbers)

*Decision: Which pages should a content/SEO team refresh first this week, out of thousands
of candidates?

Who acts: A content strategist or SEO analyst with limited weekly capacity (they can
realistically review/refresh maybe 50 pages a week, not 30,000).

Action: The output is a ranked "refresh queue" — a short list of the highest-priority
declining pages, with reason codes, that the team pulls into their sprint.

Cost of a wrong call:
- False positive (flagged as declining/urgent, but actually fine): wastes a content
  writer's time reviewing/rewriting a page that didn't need it — a few hours lost per
  page, and opportunity cost of not fixing a real problem elsewhere.
- False negative (a genuinely declining page not flagged): the page keeps losing
  traffic/rankings for weeks before anyone notices, which can mean real, compounding
  lost organic traffic for a client.

Why data/ML helps: With 30,000+ pages, a human can't manually review each one. A
hand-written rule (the baseline) is transparent but crude; an ML model can combine
many weak signals (44 columns of behavior/performance data) into a single ranked list
that's demonstrably better at surfacing the right pages first.*

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
import pandas as pd

df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

print("Total rows:", len(df))
print("Columns:", df.shape[1])

if "trend_direction" in df.columns:
    print(df["trend_direction"].value_counts())
    declining_share = (df["trend_direction"] == "down").mean()
    print(f"Share of pages currently declining: {declining_share:.1%}")
```
